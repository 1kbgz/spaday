# spaday in JupyterLite

The kernel is Python compiled to WebAssembly (Pyodide), and the widget bundles the spaday
runtime and its wasm core — a spaday UI renders in the notebook with no server. Run the
cells in order.

```python
%pip install -q spaday anywidget
```

```python
from spaday import Widget, element
from spaday.components.shell import Row, Stack


def tree(count):
    return Stack(
        element("h2").text("spaday in JupyterLite"),
        Row(element("strong").text(str(count)), element("em").text("clicks")),
    )


w = Widget(tree(0))
w  # display it: the tree renders as live spa-* web components
```

```python
# replace the tree; the browser applies a minimal diff to the live DOM
w.update(tree(1))
```

Bindings, two-way state, and `SendPatch` intents round-trip with the kernel too — see the
[notebook guide](https://1kbgz.github.io/spaday/notebook.html) for `Widget(state=...)`,
`on_state`, and `on_intent`.
